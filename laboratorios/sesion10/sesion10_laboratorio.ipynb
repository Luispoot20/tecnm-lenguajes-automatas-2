````python
%%writefile api_sesion10.h
/* api_sesion10.h - contrato de la Sesion 10
 * Mini-evaluador de expresiones (Hito 1)
 *
 * Esta cabecera es el contrato: el harness y tu implementacion
 * deben incluirla. Si cambias una firma, el harness dejara de compilar.
 */
#ifndef API_SESION10_H
#define API_SESION10_H

long evaluar_expresion(const char *fuente, int *error);

#endif /* API_SESION10_H */
````

````python
%%writefile artesano_sesion10_c.c
/* artesano_sesion10_c.c - Fase 1: El Artesano (sin IA)
 * Mini-evaluador de expresiones (Hito 1).
 *
 * Implementa lo que declara api_sesion10.h.
 * El harness compila este archivo junto con el suyo: no escribas main().
 */
#include "api_sesion10.h"
#include <stdio.h>
#include <string.h>

#include <ctype.h>

typedef struct {
    const char *s;
    int error;
} Parser;

static void saltar(Parser *p) {
    while (*p->s == ' ') {
        p->s++;
    }
}

static long expr(Parser *p);

static long primaria(Parser *p) {
    long valor = 0;

    saltar(p);

    if (p->error != 0) {
        return 0;
    }

    if (*p->s == '(') {
        p->s++;
        valor = expr(p);
        saltar(p);

        if (p->error == 0) {
            if (*p->s == ')') {
                p->s++;
            } else {
                p->error = 2;
            }
        }

        return valor;
    }

    if (isdigit((unsigned char)*p->s)) {
        while (isdigit((unsigned char)*p->s)) {
            valor = valor * 10 + (long)(*p->s - '0');
            p->s++;
        }
        return valor;
    }

    p->error = 2;
    return 0;
}

static long termino(Parser *p) {
    long valor = primaria(p);

    while (p->error == 0) {
        char operador;
        long derecho;

        saltar(p);
        operador = *p->s;

        if (operador != '*' && operador != '/') {
            break;
        }

        p->s++;
        derecho = primaria(p);

        if (p->error != 0) {
            return 0;
        }

        if (operador == '*') {
            valor *= derecho;
        } else {
            if (derecho == 0) {
                p->error = 1;
                return 0;
            }
            valor /= derecho;
        }
    }

    return valor;
}

static long expr(Parser *p) {
    long valor = termino(p);

    while (p->error == 0) {
        char operador;
        long derecho;

        saltar(p);
        operador = *p->s;

        if (operador != '+' && operador != '-') {
            break;
        }

        p->s++;
        derecho = termino(p);

        if (p->error != 0) {
            return 0;
        }

        if (operador == '+') {
            valor += derecho;
        } else {
            valor -= derecho;
        }
    }

    return valor;
}

long evaluar_expresion(const char *fuente, int *error) {
    static int error_local;
    Parser p;
    long resultado;

    if (error == NULL) {
        error = &error_local;
    }

    *error = 0;

    if (fuente == NULL) {
        *error = 2;
        return 0;
    }

    p.s = fuente;
    p.error = 0;

    resultado = expr(&p);
    saltar(&p);

    if (p.error == 0 && *p.s != '\0') {
        p.error = 2;
    }

    *error = p.error;
    return (p.error == 0) ? resultado : 0;
}

````

## Fase 2 · El Científico (10 min)

Redacta el contrato **C-R-E-O** (Contexto, Rol, Especificación, Output) para pedirle
la misma función a la IA, y pega el resultado en la celda de `ia_sesion10_c.c`. La
firma debe ser idéntica a la del contrato; si la IA la cambia, corrígela antes de
compilar.


````python
%%writefile ia_sesion10_c.c
/* ia_sesion10_c.c - Fase 2: El Cientifico
 *
 * Pega aqui el codigo que te devolvio la IA, con la MISMA firma
 * que declara api_sesion10.h. No puede traer main().
 */

#include "api_sesion10.h"
#include <stddef.h>
#include <ctype.h>

typedef struct {
    const char *actual;
    int codigo_error;
} Cursor;

static void omitir_espacios(Cursor *c) {
    while (*c->actual == ' ') {
        ++c->actual;
    }
}

static long analizar_suma(Cursor *c);

static long analizar_primario(Cursor *c) {
    long numero = 0;

    omitir_espacios(c);

    if (c->codigo_error != 0) {
        return 0;
    }

    if (*c->actual == '(') {
        long interno;
        ++c->actual;
        interno = analizar_suma(c);
        omitir_espacios(c);

        if (c->codigo_error == 0) {
            if (*c->actual != ')') {
                c->codigo_error = 2;
            } else {
                ++c->actual;
            }
        }
        return interno;
    }

    if (!isdigit((unsigned char)*c->actual)) {
        c->codigo_error = 2;
        return 0;
    }

    do {
        numero = numero * 10L + (long)(*c->actual - '0');
        ++c->actual;
    } while (isdigit((unsigned char)*c->actual));

    return numero;
}

static long analizar_producto(Cursor *c) {
    long acumulado = analizar_primario(c);

    for (;;) {
        char signo;
        long operando;

        if (c->codigo_error != 0) {
            return 0;
        }

        omitir_espacios(c);
        signo = *c->actual;

        if (signo != '*' && signo != '/') {
            return acumulado;
        }

        ++c->actual;
        operando = analizar_primario(c);

        if (c->codigo_error != 0) {
            return 0;
        }

        if (signo == '/') {
            if (operando == 0) {
                c->codigo_error = 1;
                return 0;
            }
            acumulado /= operando;
        } else {
            acumulado *= operando;
        }
    }
}

static long analizar_suma(Cursor *c) {
    long acumulado = analizar_producto(c);

    for (;;) {
        char signo;
        long operando;

        if (c->codigo_error != 0) {
            return 0;
        }

        omitir_espacios(c);
        signo = *c->actual;

        if (signo != '+' && signo != '-') {
            return acumulado;
        }

        ++c->actual;
        operando = analizar_producto(c);

        if (c->codigo_error != 0) {
            return 0;
        }

        if (signo == '+') {
            acumulado += operando;
        } else {
            acumulado -= operando;
        }
    }
}

long evaluar_expresion(const char *fuente, int *error) {
    static int error_respaldo;
    Cursor c;
    long valor;

    if (error == NULL) {
        error = &error_respaldo;
    }
    *error = 0;

    if (fuente == NULL) {
        *error = 2;
        return 0;
    }

    c.actual = fuente;
    c.codigo_error = 0;

    valor = analizar_suma(&c);
    omitir_espacios(&c);

    if (c.codigo_error == 0 && *c.actual != '\0') {
        c.codigo_error = 2;
    }

    *error = c.codigo_error;
    return (c.codigo_error == 0) ? valor : 0;
}

````

## Fase 3 · El Auditor (10 min)

El harness de abajo **no aprueba nada por sí solo**: ejecuta tu función caso por caso,
la compara contra el valor esperado e imprime `OK` o `FALLA`. Léelo antes de ejecutarlo:
los casos te dicen qué se está midiendo.


````python
%%writefile harness_sesion10_c.c
/* harness_sesion10_c.c - auditoria automatica de la Sesion 10
 *
 * Compilacion:
 *   gcc -std=c11 -Wall -Wextra -Werror harness_sesion10_c.c artesano_sesion10_c.c -o test_artesano
 *
 * El harness NO aprueba nada por si solo: cada linea se calcula
 * ejecutando tu implementacion y comparandola contra el valor esperado.
 * Si algun caso falla, el programa termina con codigo de salida 1.
 */
#include <stdio.h>
#include <string.h>
#include "api_sesion10.h"

static int probadas = 0;
static int fallidas = 0;

static void verificar_int(const char *caso, long esperado, long obtenido) {
    int ok = (esperado == obtenido);
    probadas++;
    if (!ok) fallidas++;
    printf("[%2d] %-34s esperado: %-12ld obtenido: %-12ld %s\n",
           probadas, caso, esperado, obtenido, ok ? "OK" : "FALLA");
}

int main(void) {
    printf("=== AUDITORIA SESION 10 - Mini-evaluador de expresiones (Hito 1) ===\n");
    printf("Caso                               esperado:              obtenido:              resultado\n");

    { int e=0; long r=evaluar_expresion("2+3*4",&e); (void)e; verificar_int("\"2+3*4\"", (long)(14), (long)(r)); }
    { int e=0; long r=evaluar_expresion("(2+3)*4",&e); (void)e; verificar_int("\"(2+3)*4\"", (long)(20), (long)(r)); }
    { int e=0; long r=evaluar_expresion("10/0",&e); (void)r; verificar_int("\"10/0\" -> error 1", (long)(1), (long)(e)); }
    { int e=0; long r=evaluar_expresion("2+",&e); (void)r; verificar_int("\"2+\" -> error 2", (long)(2), (long)(e)); }
    { int e=0; long r=evaluar_expresion("  7 ",&e); (void)e; verificar_int("\"  7 \"", (long)(7), (long)(r)); }
    { int e=0; long r=evaluar_expresion("5-10",&e); (void)e; verificar_int("\"5-10\"", (long)(-5), (long)(r)); }

    printf("\nRESULTADO: %d/%d pruebas superadas\n",
           probadas - fallidas, probadas);
    if (fallidas != 0) {
        printf("Aun hay %d caso(s) sin resolver. Revisa tu implementacion.\n",
               fallidas);
        return 1;
    }
    printf("TODO EN VERDE. Documenta tu auditoria en auditorias/auditoria_sesion10.md" "\n");
    return 0;
}
````
